## 1. Tuple

A tuple is an ordered collection that cannot be changed after creation.

```python
point = (10, 20)
x, y = point
print(x, y)
```

## 2. Set

A set stores unique values and is useful for membership and set operations.

```python
values = {1,2,2,3}
print(values)
print({1,2} & {2,3})
```

## 3. Dictionary

A dictionary maps keys to values.

```python
student = {"name":"Renco", "score":90}
print(student["name"])
student["score"] = 95
```

## 4. Safe dictionary access

get() can provide a default instead of raising a KeyError.

```python
email = student.get("email", "Not provided")
print(email)
```
