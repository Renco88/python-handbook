## 1. Indexing

A string is a sequence, so each character has a position.

```python
word = "Python"
print(word[0])
print(word[-1])
```

## 2. Slicing

Slicing extracts a range using start:stop:step.

```python
word = "Python"
print(word[0:3])
print(word[::-1])
```

## 3. String methods

Methods perform common text operations.

```python
name = "  mahadi hassan  "
print(name.strip().title())
print(name.replace("hassan", "renco"))
```

## 4. f-strings

f-strings make formatted output readable.

```python
name = "Renco"
score = 92
print(f"{name} scored {score}%")
```
