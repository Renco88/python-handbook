## NumPy

```python
import numpy as np
values = np.array([10, 20, 30])
print(values.mean())
```

## Pandas

```python
import pandas as pd
df = pd.DataFrame({"name":["A","B"],"score":[80,90]})
print(df[df["score"] >= 85])
```

## Matplotlib

```python
import matplotlib.pyplot as plt
plt.plot([1,2,3],[2,5,4])
plt.title("Simple chart")
plt.show()
```
