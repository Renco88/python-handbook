## Write a file

```python
from pathlib import Path
Path("notes.txt").write_text("Hello Python!", encoding="utf-8")
```

## Read a file

```python
from pathlib import Path
text = Path("notes.txt").read_text(encoding="utf-8")
print(text)
```

## JSON

```python
import json
data = {"name": "Renco", "skills": ["Python", "React"]}
text = json.dumps(data, indent=2)
print(text)
```
