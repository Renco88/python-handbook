## 1. Defining a function

Functions package reusable behavior.

```python
def add(a, b):
    return a + b

print(add(10, 20))
```

## 2. Default and keyword arguments

Defaults make arguments optional and keyword arguments make calls clearer.

```python
def greet(name, message="Welcome"):
    return f"{message}, {name}!"
print(greet("Renco"))
print(greet("Renco", message="Good morning"))
```

## 3. *args and **kwargs

They collect variable-length positional and keyword arguments.

```python
def report(*scores, **info):
    print(scores)
    print(info)
report(80,90,name="Renco")
```

## 4. Recursion

A recursive function calls itself and needs a base case.

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
```
