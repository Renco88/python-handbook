## 1. if / else

Conditional statements let a program choose a path.

```python
marks = 82
if marks >= 80:
    print("A+")
else:
    print("Keep practicing")
```

## 2. elif

elif handles additional alternatives.

```python
marks = 72
if marks >= 80:
    grade = "A+"
elif marks >= 70:
    grade = "A"
else:
    grade = "Below A"
print(grade)
```

## 3. Nested decisions

A condition can contain another condition.

```python
username = "admin"
password = "123"
if username == "admin":
    if password == "123":
        print("Access granted")
```

## 4. Truthiness

Many values can be treated as true or false in a condition.

```python
items = []
if not items:
    print("The list is empty")
```
