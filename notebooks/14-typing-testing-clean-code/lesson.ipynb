## 1. Type hints

Type hints document intended types and help tools detect mistakes.

```python
def add(a: int, b: int) -> int:
    return a + b
```

## 2. Assertions

Assertions check assumptions during development and tests.

```python
def add(a,b): return a+b
assert add(2,3) == 5
```

## 3. Logging

Logging records useful runtime information.

```python
import logging
logging.basicConfig(level=logging.INFO)
logging.info("Application started")
```
