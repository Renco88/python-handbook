## List comprehension

```python
squares = [n * n for n in range(1, 6)]
print(squares)
```

## Generator

```python
def countdown(n):
    while n:
        yield n
        n -= 1
print(list(countdown(5)))
```

## Decorator

```python
def log_call(fn):
    def wrapper(*args, **kwargs):
        print("Calling", fn.__name__)
        return fn(*args, **kwargs)
    return wrapper

@log_call
def hello():
    return "Hi"
print(hello())
```
