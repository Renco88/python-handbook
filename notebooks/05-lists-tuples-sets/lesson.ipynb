# Chapter 05: Lists, Tuples & Sets

Understand Python collections, when to use each one, common methods, unpacking and practical data handling.

Run every example and modify the values yourself.

## List operations

```python
fruits = ["apple", "banana"]
fruits.append("mango")
fruits[0] = "orange"
print(fruits)
```

## Tuple unpacking

```python
point = (10, 20)
x, y = point
print(x, y)
```

## Set operations

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)
print(a & b)
```
