## 1. Create a database

SQLite stores structured data in a local file.

```python
import sqlite3
con=sqlite3.connect("app.db")
cur=con.cursor()
cur.execute("CREATE TABLE IF NOT EXISTS users(id INTEGER PRIMARY KEY,name TEXT)")
con.commit()
```

## 2. Insert safely

Use parameters rather than string concatenation.

```python
cur.execute("INSERT INTO users(name) VALUES (?)", ("Renco",))
con.commit()
```

## 3. Read data

SELECT returns rows that you can fetch from Python.

```python
rows=cur.execute("SELECT id,name FROM users").fetchall()
print(rows)
```

## 4. CRUD mindset

Create, Read, Update and Delete cover the basic lifecycle of stored records.

```python
cur.execute("UPDATE users SET name=? WHERE id=?", ("Mahadi",1))
con.commit()
```
