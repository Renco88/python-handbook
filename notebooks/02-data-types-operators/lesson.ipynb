## 1. Integers and floats

int represents whole numbers; float represents numbers with decimal parts.

```python
a = 10
b = 2.5
print(a + b)
print(type(b))
```

## 2. Arithmetic operators

Arithmetic operators calculate values: +, -, *, /, //, %, **.

```python
x = 17
y = 5
print(x + y)
print(x // y)
print(x % y)
print(x ** 2)
```

## 3. Comparison

Comparison operators return True or False.

```python
age = 20
print(age >= 18)
print(age == 20)
```

## 4. Logical operators

and, or and not combine Boolean conditions.

```python
age = 20
has_id = True
print(age >= 18 and has_id)
```
