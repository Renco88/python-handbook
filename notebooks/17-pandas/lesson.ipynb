## 1. DataFrame

A DataFrame is a labeled two-dimensional table.

```python
import pandas as pd
df=pd.DataFrame({"name":["A","B"],"score":[80,90]})
print(df)
```

## 2. Selection

Use column names and boolean masks to select data.

```python
print(df["score"])
print(df[df["score"] >= 85])
```

## 3. GroupBy

groupby lets you calculate summaries by category.

```python
df=pd.DataFrame({"dept":["CSE","CSE","EEE"],"score":[80,90,70]})
print(df.groupby("dept")["score"].mean())
```

## 4. Missing data

Real datasets often contain missing values that need explicit handling.

```python
df=df.fillna(0)
print(df)
```
