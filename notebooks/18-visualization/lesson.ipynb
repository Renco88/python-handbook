# Chapter 18: Visualization

Matplotlib line/scatter/bar/histogram, labels, legends, subplots and storytelling

Run each example, change a value, and observe the result.

## 1. Line plot

Line plots show trends across an ordered x-axis.

```python
import matplotlib.pyplot as plt
plt.plot([1,2,3,4],[2,4,3,6])
plt.xlabel("Day")
plt.ylabel("Value")
plt.show()
```

## 2. Scatter plot

Scatter plots show relationships between two numeric variables.

```python
plt.scatter([1,2,3,4],[2,5,3,7])
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## 3. Histogram

Histograms show the distribution of numeric values.

```python
plt.hist([1,2,2,3,3,3,4,5], bins=5)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## 4. Subplots

Multiple plots can tell a fuller story in one figure.

```python
fig,ax=plt.subplots(1,2)
ax[0].plot([1,2,3],[2,4,3])
ax[1].scatter([1,2,3],[3,1,4])
plt.show()
```
