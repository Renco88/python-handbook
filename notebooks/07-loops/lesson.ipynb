## 1. for loop

for repeats a block for each item in an iterable.

```python
for name in ["A", "B", "C"]:
    print(name)
```

## 2. range()

range generates a sequence of numbers without creating a large list.

```python
for i in range(1, 6):
    print(i)
```

## 3. enumerate() and zip()

enumerate gives an index; zip walks through multiple sequences together.

```python
names=["A","B"]
scores=[80,90]
for i,(name,score) in enumerate(zip(names,scores), start=1):
    print(i,name,score)
```

## 4. break and continue

break stops a loop; continue skips the current iteration.

```python
for n in range(1,6):
    if n == 3:
        continue
    if n == 5:
        break
    print(n)
```
