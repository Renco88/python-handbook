## 1. Unpacking

Python can unpack sequences into separate variables.

```python
first, *middle, last = [1,2,3,4]
print(first, middle, last)
```

## 2. Generator

Generators produce values lazily with yield.

```python
def countdown(n):
    while n:
        yield n
        n -= 1
print(list(countdown(3)))
```

## 3. Decorator

A decorator wraps a function to add behavior.

```python
def log_call(fn):
    def wrapper(*args, **kwargs):
        print("Calling", fn.__name__)
        return fn(*args, **kwargs)
    return wrapper

@log_call
def hello(): return "Hi"
print(hello())
```

## 4. Context manager

with is a clean way to acquire and release resources.

```python
with open("notes.txt", encoding="utf-8") as f:
    text=f.read()
```
