## Indexing and slicing

```python
text = "Python"
print(text[0])
print(text[-1])
print(text[1:4])
print(text[::-1])
```

## Useful methods

```python
name = "  mahadi hassan  "
print(name.strip().title())
print(name.replace("hassan", "renco"))
```

## f-string

```python
product = "Laptop"
price = 65000
print(f"{product} costs ৳{price:,}")
```
