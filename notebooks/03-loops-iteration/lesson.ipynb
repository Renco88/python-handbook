Master for, while, range, break, continue, nested loops and practical repetition patterns.

Run every example and modify the values yourself.

## for loop

```python
for i in range(1, 6):
    print(i)
```

## while loop

```python
count = 5
while count > 0:
    print(count)
    count -= 1
```

## break / continue

```python
for n in range(1, 11):
    if n == 7:
        break
    if n % 2 == 0:
        continue
    print(n)
```
