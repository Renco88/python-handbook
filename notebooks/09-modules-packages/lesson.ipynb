## 1. Importing modules

Modules let you reuse code from another file or the standard library.

```python
import math
print(math.sqrt(144))
```

## 2. Creating a module

Your own .py file can be imported by another script.

```python
# tools.py
def double(x):
    return x * 2

# main.py
from tools import double
print(double(8))
```

## 3. pip and environments

Virtual environments isolate project dependencies.

```python
python -m venv .venv
# Windows PowerShell
.venv\Scripts\Activate.ps1
pip install requests
```
