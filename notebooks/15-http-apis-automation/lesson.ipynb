## 1. HTTP mental model

A client sends a request; a server returns a response with a status code and data.

```python
import requests
r=requests.get("https://api.github.com", timeout=10)
print(r.status_code)
```

## 2. JSON API

Many APIs return JSON that becomes Python dictionaries/lists.

```python
data=r.json()
print(type(data))
print(data.keys())
```

## 3. POST request

POST commonly sends data to a server.

```python
payload={"name":"Renco"}
r=requests.post("https://httpbin.org/post", json=payload, timeout=10)
print(r.status_code)
```

## 4. Automation

Automation means turning a repeated manual workflow into deterministic code.

```python
from pathlib import Path
for p in Path(".").glob("*.txt"):
    print("Found:",p.name)
```
