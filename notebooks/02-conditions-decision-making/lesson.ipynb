# Chapter 02: Conditions & Decision Making

Teach Python how to make decisions with if, elif, else, comparisons, logical operators and nested conditions.

Run every example and modify the values yourself.

## Simple decision

```python
age = 20
if age >= 18:
    print("Adult")
else:
    print("Minor")
```

## Multiple branches

```python
marks = 82
if marks >= 80:
    grade = "A+"
elif marks >= 70:
    grade = "A"
else:
    grade = "Below A"
print(grade)
```

## Logical operators

```python
age = 22
has_id = True
if age >= 18 and has_id:
    print("Entry allowed")
```
