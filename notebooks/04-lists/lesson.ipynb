## 1. Creating a list

Lists store multiple values in one ordered, mutable collection.

```python
fruits = ["apple", "banana", "mango"]
print(fruits[1])
```

## 2. List methods

append adds, insert places, remove deletes by value, and pop removes and returns an item.

```python
items = [10, 20, 30]
items.append(40)
items.remove(20)
last = items.pop()
print(items, last)
```

## 3. Nested lists

A list can contain other lists, which is useful for tables and matrices.

```python
matrix = [[1, 2], [3, 4]]
print(matrix[1][0])
```

## 4. List comprehension

A comprehension creates a new list from an iterable with optional filtering.

```python
numbers = [1,2,3,4,5]
squares = [n*n for n in numbers if n % 2 == 0]
print(squares)
```
