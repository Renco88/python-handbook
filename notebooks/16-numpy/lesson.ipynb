## 1. Array vs Python list

NumPy arrays are designed for efficient numerical operations.

```python
import numpy as np
a=np.array([1,2,3])
print(a*2)
```

## 2. Indexing and slicing

NumPy supports multidimensional indexing and slicing.

```python
a=np.arange(12).reshape(3,4)
print(a[1,2])
print(a[:,1])
```

## 3. Aggregation

Functions such as mean, sum, min and max summarize arrays.

```python
a=np.array([10,20,30,40])
print(a.mean(), a.sum(), a.max())
```

## 4. Masks

Boolean masks select values matching a condition.

```python
a=np.array([10,15,20,25])
print(a[a >= 20])
```
