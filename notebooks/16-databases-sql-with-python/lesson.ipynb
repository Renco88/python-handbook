## SQLite

```python
import sqlite3
con = sqlite3.connect("app.db")
cur = con.cursor()
cur.execute("CREATE TABLE IF NOT EXISTS users (id INTEGER PRIMARY KEY, name TEXT)")
cur.execute("INSERT INTO users(name) VALUES (?)", ("Renco",))
con.commit()
print(cur.execute("SELECT * FROM users").fetchall())
con.close()
```
