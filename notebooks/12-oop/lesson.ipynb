## 1. Class and object

A class is a blueprint; an object is an instance created from it.

```python
class Student:
    def __init__(self, name):
        self.name = name

s = Student("Renco")
print(s.name)
```

## 2. Instance method

Methods operate on object data.

```python
class Student:
    def __init__(self,name): self.name=name
    def introduce(self): return f"I am {self.name}"
print(Student("Renco").introduce())
```

## 3. Inheritance and overriding

A child class can inherit and replace behavior from a parent.

```python
class Animal:
    def speak(self): return "sound"
class Dog(Animal):
    def speak(self): return "woof"
print(Dog().speak())
```

## 4. Encapsulation and abstraction

Internal details can be protected behind methods; abstract classes define required behavior.

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self): pass
class Square(Shape):
    def __init__(self,s): self.s=s
    def area(self): return self.s*self.s
print(Square(4).area())
```
