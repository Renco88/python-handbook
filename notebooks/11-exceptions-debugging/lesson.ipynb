## 1. Reading a traceback

A traceback tells you where an exception happened and what type it was.

```python
numbers = [1,2]
print(numbers[5])
```

## 2. try / except

Handle expected runtime problems without crashing the whole program.

```python
try:
    n=int(input("Number: "))
    print(10/n)
except ValueError:
    print("Enter an integer")
except ZeroDivisionError:
    print("Zero is not allowed")
```

## 3. raise

raise lets your program reject invalid state explicitly.

```python
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("Insufficient balance")
    return balance-amount
```

## 4. finally

finally runs whether an exception happened or not.

```python
try:
    print(10/2)
finally:
    print("Cleanup")
```
