# Chapter 07: Functions

Build reusable programs with parameters, return values, scope, default arguments, *args, **kwargs, lambda and recursion.

Run every example and modify the values yourself.

## Return value

```python
def add(a, b):
    return a + b
print(add(10, 20))
```

## Default argument

```python
def greet(name, message="Welcome"):
    print(f"{message}, {name}!")
greet("Renco")
greet("Renco", "Good morning")
```

## args and kwargs

```python
def report(*scores, **info):
    print("scores:", scores)
    print("info:", info)
report(80, 90, name="Renco")
```
