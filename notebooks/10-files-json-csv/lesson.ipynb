## 1. Writing a file

open() with mode w creates or overwrites a file.

```python
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("Hello Python")
```

## 2. Reading a file

read() loads the contents as text.

```python
with open("notes.txt", "r", encoding="utf-8") as f:
    content = f.read()
print(content)
```

## 3. JSON

JSON represents structured data and maps naturally to Python dictionaries and lists.

```python
import json
data={"name":"Renco","skills":["Python","React"]}
text=json.dumps(data, indent=2)
print(text)
```

## 4. CSV

CSV stores tabular data as rows and columns.

```python
import csv
with open("students.csv", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        print(row["name"])
```
