## 1. I/O-bound vs CPU-bound

Waiting for network/files is different from heavy computation. This distinction guides concurrency choices.

```python
import time
print("Start")
time.sleep(1)
print("Done")
```

## 2. Async task

async def defines a coroutine and await pauses it without blocking other tasks.

```python
import asyncio
async def task(name):
    await asyncio.sleep(1)
    print(name)
async def main():
    await asyncio.gather(task("A"),task("B"))
asyncio.run(main())
```

## 3. Threading concept

Threads can be useful for I/O-bound work.

```python
from concurrent.futures import ThreadPoolExecutor
with ThreadPoolExecutor(max_workers=2) as ex:
    print(list(ex.map(lambda x:x*x,[1,2,3])))
```

## 4. Multiprocessing concept

Separate processes can use multiple CPU cores for CPU-heavy work.

```python
from concurrent.futures import ProcessPoolExecutor
with ProcessPoolExecutor() as ex:
    print(list(ex.map(lambda x:x*x,[1,2,3])))
```
