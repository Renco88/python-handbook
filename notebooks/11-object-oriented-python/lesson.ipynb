## Class and object

```python
class Student:
    def __init__(self, name):
        self.name = name
    def introduce(self):
        return f"I am {self.name}."

s = Student("Renco")
print(s.introduce())
```

## Inheritance

```python
class Animal:
    def speak(self):
        print("Some sound")
class Dog(Animal):
    def speak(self):
        print("Woof!")
Dog().speak()
```
