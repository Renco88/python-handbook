# Chapter 06: Dictionaries

Learn key-value data, nested dictionaries, looping, safe access, updates and real-world records.

Run every example and modify the values yourself.

## Basic dictionary

```python
student = {"name": "Renco", "age": 25, "dept": "CSE"}
print(student["name"])
student["age"] = 26
print(student)
```

## Safe access

```python
email = student.get("email", "Not provided")
print(email)
```

## Nested data

```python
users = {"renco": {"role": "developer", "active": True}}
print(users["renco"]["role"])
```
