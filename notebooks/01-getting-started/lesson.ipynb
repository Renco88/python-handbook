## 1. What is Python?

Python is a high-level programming language designed to make programs readable.

```python
print("Hello, Python!")
```

## 2. Variables

A variable is a name that refers to a value. Think of it as a label attached to information.

```python
name = "Mahadi"
age = 25
print(name)
print(age)
```

## 3. Input

input() lets a program receive text from the user.

```python
name = input("What is your name? ")
print(f"Hello, {name}!")
```

## 4. Type conversion

User input is text, so numeric calculations usually need int() or float().

```python
age = int(input("Age: "))
print(age + 1)
```
